Un objetivo central de la descripción de datos es decidir si los datos son
una muestra homogénea de una población o corresponden a una
mezcla de poblaciones distintas que deben estudiarse
separadamente.
Un caso de heterogeneidad es la presencia de una pequeña proporción
de observaciones atípicas (outliers), que corresponden a datos
heterogéneos con el resto. El coeficiente de curtosis puede ayudar en este objetivo, ya que tomará
un valor alto, mayor que 7 u 8. Un valor alto de la curtosis para una
variable implica heterogeneidad por uno pocos atípicos muy alejados
del resto.

Si hay unos pocos datos atípicos muy alejados del resto, la variabilidad
de las desviaciones será grande, debido a estos valores y los coeficientes
de curtosis o de homogeneidad serán altos.

Un tipo distinto de heterogeneidad es cuando se tiene una
mezcla de dos poblaciones, de manera que una proporción
importante de los datos, entre el 25% y el 50%, son
heterogéneos con el resto. En este caso, el coeficiente de
curtosis es pequeño, menor de dos, y es fácil comprobar que si
se mezcla a partes iguales dos distribuciones muy distintas, la
curtosis de la distribución resultante tiende a uno, el valor
mínimo del coeficiente, cuando aumenta la separación entre
las poblaciones.

### Análisis de una variable angular: `Solar Zenith Angle (°)`

Se seleccionó el **ángulo cenital solar** de `df_I`. La curtosis calculada por Pandas es **−0.231** y corresponde a la curtosis de Fisher (exceso de curtosis). Para compararla con el criterio explicado arriba se suma 3, obteniendo una curtosis de Pearson de **2.769**. Este valor está muy por debajo de 7 u 8, por lo que la curtosis **no indica la presencia de unos pocos valores atípicos muy alejados del resto**. Tampoco es menor que 2, así que no aporta evidencia clara de una mezcla de dos poblaciones muy separadas.

Como comprobación adicional se aplicó el criterio del rango intercuartílico: $Q_1 = 46.239°$, $Q_3 = 74.248°$ e $IQR = 28.010°$. Los límites son **4.224°** y **116.263°**. Con este método aparecen **8 observaciones atípicas de 1,000 (0.8 %)**, todas por debajo del límite inferior.

La variable presenta un coeficiente de asimetría de **−0.430**. El signo negativo indica una **asimetría ligera hacia la izquierda**: la cola de la distribución se extiende hacia los ángulos más bajos. Esto concuerda con el resultado del IQR, ya que las 8 observaciones señaladas como atípicas se encuentran precisamente en el extremo inferior. Como la magnitud del coeficiente es menor que 0.5, la asimetría es leve y la distribución puede considerarse aproximadamente simétrica.

**Conclusión:** el método IQR sí identifica una proporción muy pequeña de atípicos en la cola izquierda. Sin embargo, la asimetría es leve y la curtosis muestra que no son valores extremos suficientemente alejados como para producir una heterogeneidad fuerte. Además, los ángulos cercanos a 0° son físicamente posibles cuando el Sol está próximo al cenit; por tanto, estos datos no deben considerarse errores ni eliminarse automáticamente.

### Análisis de una variable angular de `df_II`: `Topocentric zenith angle`

Se seleccionó el **ángulo cenital topocéntrico**. El código calcula el **coeficiente de curtosis de Pearson** mediante `numeric_df_II.kurt() + 3`, que es la medida adecuada para aplicar los umbrales del texto. Su valor es **2.051**. El mínimo observado es **14.967°** y el máximo **164.994°**, ambos dentro de esos límites; por tanto, **no se detectaron valores atípicos entre las 105,216 observaciones**.

**Conclusión:** el ángulo cenital topocéntrico de `df_II` presenta una distribución aproximadamente simétrica y aplanada, sin valores atípicos según el criterio IQR y sin señales de observaciones extremas aisladas según la curtosis.
